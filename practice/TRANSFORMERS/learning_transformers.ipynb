```python
import keras 

filename = keras.utils.get_file(
    origin = (
    "https://storage.googleapis.com/download.tensorflow.org/"
        "data/shakespeare.txt"
    ),
)

shakespeare = open(filename, "r").read()
```

```python
# What does it contain? 
print(shakespeare[:250])
```

```python
# To build a language model from this input, we need to split our data into equal length chunks that we can batch and use for model training. 

import tensorflow as tf

# the chunk size wwe will use during training. We only train on sequences of 100 characters at a time 

sequence_lenght = 100

# Generator function: 
# meaning a function that can produce values on demand. 
def split_input(input, sequence_lenght): 
    for i in range(0, len(input), sequence_lenght): 
        yield input[i: i + sequence_lenght]

""" 
Small example: 
split_input(....) creates an generator object 
print(....) displays that generator on screen. 
print returns None and that's what get stored in g. The generator itself is thrown away. 

g = print(split_input("abcdefghij", 4))
print(g)
print(next(g))
"""
        
features = list(split_input(shakespeare[:-1], sequence_lenght))
labels = list(split_input(shakespeare[1:], sequence_lenght))
dataset = tf.data.Dataset.from_tensor_slices((features, labels))

```
